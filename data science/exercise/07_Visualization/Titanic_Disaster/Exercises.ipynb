### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

%matplotlib inline
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Titanic_Desaster/train.csv)

### Step 3. Assign it to a variable titanic 

```python
url = 'train.csv'
t = pd.read_csv(url)
t.head()
```

### Step 4. Set PassengerId as the index 

```python
t.set_index('PassengerId', inplace=True)
t.head()
```

### Step 5. Create a pie chart presenting the male/female proportion

```python
males = (t.Sex == 'male').sum()
females = (t.Sex == 'female').sum()

plt.pie(
	[males, females],
	labels=['Males', 'Females'],
	autopct = '%1.1f%%'
)
```

### Step 6. Create a scatterplot with the Fare payed and the Age, differ the plot color by gender

```python
plt.scatter(
  t.Age, t.Fare, c=[ 'red' if sex == 'male' else 'blue' for sex in t.Sex ]
)
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Scatter plot of Age vs Fare colored by Sex')
```

### Step 7. How many people survived?

```python
un_RIP_TvT = (t.Survived == 1).sum()
un_RIP_TvT
```

### Step 8. Create a histogram with the Fare payed

```python
t1 = t.Fare.sort_values(ascending=False)
interval_range = np.arange(0, 600, 10)
plt.hist(t1, bins=interval_range)
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.title('Fare Paid Histogram')
```
