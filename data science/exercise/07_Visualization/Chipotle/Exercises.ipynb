### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# set this so the graphs open internally
%matplotlib inline
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called chipo.

```python
url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv'
c = pd.read_csv(url, sep='\t')
```

### Step 4. See the first 10 entries

```python
c.head(10)
```

### Step 5. Create a histogram of the top 5 items bought

```python
t = c.item_name
letter_cnt = Counter(t)
df = pd.DataFrame.from_dict(letter_cnt, orient='index')
df.rename(columns={0: 'Frequency'}, inplace=True)
df = df['Frequency'].sort_values(ascending=True)[45:50]
df.plot(kind='bar')
```

### Step 6. Create a scatterplot with the number of items orderered per order price
#### Hint: Price should be in the X-axis and Items ordered in the Y-axis

```python
c.item_price = c.item_price.apply(lambda x: float(x[1:-1]))
od = c.groupby('order_id').sum()
plt.scatter(x = od.item_price, y = od.quantity, s=50, c='green')
plt.xlabel('Order Price')
plt.ylabel('Items ordered')
plt.title('Number of items ordered per order price')
plt.ylim(0)
```
